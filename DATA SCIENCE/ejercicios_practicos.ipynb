```python
import pandas as pd

# 1. Carga de datos
movies  = pd.read_csv('movies.csv')
ratings = pd.read_csv('ratings.csv')

# 2. Tamaño de los DataFrames
print("Películas:", movies.shape[0])
print("Valoraciones:", ratings.shape)

# 3. Unión de DataFrames
movie_ratings = ratings.merge(
    movies[['movieId','title','genres']],
    on='movieId',
    how='inner'
)

# 4. Valoraciones de “Dark Knight, The (2008)”
dk = movie_ratings[movie_ratings['title']=='Dark Knight, The (2008)']
print("Total valoraciones de Dark Knight:",    dk.shape[0])
print("Valoraciones de 5 estrellas:", dk[dk['rating']>==5].shape[0])

# 5. Estadísticas por película
movie_stats = movie_ratings.groupby('title').agg(
    rating_count=('rating','size'),
    rating_mean=('rating','mean')
).reset_index()
print(movie_stats)


# 6. Top 10 (al menos 100 valoraciones)
top10 = (
    movie_stats[movie_stats['rating_count']>=100]
    .sort_values('rating_mean', ascending=False)
    .head(10)
)
print(top10)


# 7. Fecha legible
ratings['date'] = pd.to_datetime(ratings['timestamp'], unit='s')
print("Primera valoración:", ratings['date'].min())
print("Última valoración:",    ratings['date'].max())
```

```python
import pandas as pd

# 1. Carga de datos con separador ';'
students = pd.read_csv('student-mat.csv', sep=';')

# 2. Dimensiones
print("Filas y columnas:", students.shape)

# 3. Filtrar menores de 18
students = students[students['age'] < 18]

# 4. Renombrar columnas de notas
students = students.rename(
    columns={'G1':'grade1','G2':'grade2','G3':'grade3'}
)

# 5. Selección de columnas
cols_keep = ['school','sex','age','activities','grade1','grade2','grade3']
students = students[cols_keep]

# 6. Columnas booleanas de aprobado (≥10)
for i in (1,2,3):
    students[f'grade{i}_passed'] = students[f'grade{i}'] >= 10

passed_all   = students[
    students[['grade1_passed','grade2_passed','grade3_passed']].all(axis=1)
].shape[0]
failed_all   = students[
    ~students[['grade1_passed','grade2_passed','grade3_passed']].any(axis=1)
].shape[0]
print("Aprobados los 3 cursos:", passed_all)
print("No aprobó ninguno:",    failed_all)

# 7. Nota media por estudiante
students['final_grade'] = students[['grade1','grade2','grade3']].mean(axis=1)

# 8. Nota media por edad y sexo
final_grade_stats = (
    students.groupby(['age','sex'], as_index=False)['final_grade']
    .mean()
    .round(2)
    .sort_values('final_grade', ascending=False)
)
print(final_grade_stats)

# 9. Detección y eliminación de duplicados
n_dup = students.duplicated().sum()
print("Filas duplicadas:", n_dup)
if n_dup > 0:
    students = students.drop_duplicates()
    print("Duplicados eliminados.")
```
